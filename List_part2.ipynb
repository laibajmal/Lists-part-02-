👥 TASK 4: REMOVE DUPLICATE NAMES
━━━━━━━━━━━━━━━━━━

You have a list of student names where some names appear more than once.

Convert the list into a set to keep only the unique names.

Expected Output:

Unique Students:
Ali
Ahmed
Sara
Usman

Note: Sets are unordered, so the order of names in your output may be different.


```python
unique_students = [ "Ali", "Ahmed", "Sara", "Usman", "Sara", "Ahmed", "Ali", "Sara"]
print("Unique Students: ", list(set(unique_students)))
```

💻 TASK 5: EMPLOYEE SKILLS
━━━━━━━━━━━━━━━━━━

Create a list of employee skills. Some skills are repeated.

Convert the list into a set to find the unique skills in the company.

Expected Output:

Unique Skills:
Python
Java
SEO
WordPress
Flutter

Note: Sets are unordered, so the order of skills in your output may be different.

```python


unique_skills = ["Python", "WordPress", "Java", "WordPress", "SEO", "WordPress", "Flutter", "WordPress", "Python", "WordPress", "Flutter"]
print("Unique Skills: ", list(set(unique_skills)))
```
